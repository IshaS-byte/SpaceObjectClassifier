```python
import numpy as np
import tensorflow as tf
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import joblib
import os
import gc
from tqdm import tqdm
```

```python
print("CLASSICAL MACHINE LEARNING MODEL")
train_dir = "data/train"
test_dir = "data/test"
IMG_SIZE = 64 
BATCH_SIZE = 32
```

```python
#datasets
print("\nLoading datasets...")
train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    train_dir,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    test_dir,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    shuffle=False
)

class_names = train_ds.class_names
num_classes = len(class_names)
print(f"Classes: {class_names}")
```

```python
def extract_features(dataset, dataset_name):
    """Extract flattened pixel features from images."""
    X = []
    y = []
    
    print(f"\nExtracting features from {dataset_name}...")
    for images, labels in dataset:
        # Normalize to 0-1
        images = images.numpy() / 255.0
        # Flatten images
        for i in range(len(images)):
            X.append(images[i].flatten())
            y.append(labels[i].numpy())
    
    X = np.array(X)
    y = np.array(y)
    
    print(f"  Shape: {X.shape}")
    print(f"  Samples: {len(y)}")
    
    return X, y
```

```python
X_train, y_train = extract_features(train_ds, "training set")
X_test, y_test = extract_features(test_ds, "test set")
gc.collect()
```

```python
print("TRAINING CLASSICAL MODELS")
models = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, n_jobs=-1),
    'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=5, n_jobs=-1),
    'Support Vector Machine': SVC(kernel='rbf', random_state=42)
}
results = {}

for name, model in models.items():
    print(f"Training {name}...")
       
    # Train
    model.fit(X_train, y_train)
    
    # Predict
    y_pred = model.predict(X_test)
    
    # Evaluate
    accuracy = accuracy_score(y_test, y_pred)
    
    print(f"\n{name} Results:")
    print(f"  Accuracy: {accuracy * 100:.2f}%")
    
    # Store results
    results[name] = {
        'model': model,
        'accuracy': accuracy,
        'predictions': y_pred
    }
    
    # Classification report
    print(f"\nClassification Report:")
    report = classification_report(
        y_test, 
        y_pred, 
        labels=list(range(num_classes)),
        target_names=class_names,
        zero_division=0
    )
    print(report)
    
    results[name]['report'] = report
    
    gc.collect()
```

```python
print("SAVING RESULTS")
os.makedirs("results", exist_ok=True)

# Save each model independently
for name, data in results.items():
    filename = f"results/classical_model_{name.lower().replace(' ', '_')}.pkl"
    joblib.dump(data['model'], filename)
    print(f"{name} saved to: {filename}")
```

```python
with open("results/classical_performance.txt", "w") as f:
    f.write("=" * 60 + "\n")
    f.write("CLASSICAL ML MODELS PERFORMANCE\n")
    f.write("=" * 60 + "\n\n")
    
    f.write("Classes:\n")
    for i, c in enumerate(class_names):
        f.write(f"  {i}: {c}\n")
    
    f.write("\n" + "=" * 60 + "\n")
    f.write("MODEL RESULTS\n")
    f.write("=" * 60 + "\n\n")
    
    for name in models.keys():
        f.write(f"\n{name}:\n")
        f.write("-" * 60 + "\n")
        f.write(f"Accuracy: {results[name]['accuracy'] * 100:.2f}%\n\n")
        f.write(results[name]['report'])
        f.write("\n")
    
    f.write("\n" + "=" * 60 + "\n")
    f.write("CONFIGURATION\n")
    f.write("=" * 60 + "\n\n")
    f.write(f"Image Size: {IMG_SIZE}x{IMG_SIZE}\n")
    f.write(f"Feature Vector Size: {X_train.shape[1]}\n")
    f.write(f"Training Samples: {len(y_train)}\n")
    f.write(f"Test Samples: {len(y_test)}\n")
    f.write(f"\nModels Trained:\n")
    for name in models.keys():
        f.write(f"  - {name}\n")


print("Performance report saved to: results/classical_performance.txt")
```

```python
print("CONFUSION MATRICES")
for name in models.keys():
    print(f"\n{name}:")
    y_pred = results[name]['predictions']
    
    for i, class_name in enumerate(class_names):
        class_mask = y_test == i
        if class_mask.sum() > 0:
            correct = ((y_test == i) & (y_pred == i)).sum()
            total = class_mask.sum()
            class_acc = correct / total * 100
            print(f"  {class_name}: {class_acc:.2f}% ({correct}/{total})")
```

```python
print("Classical Training Complete")
print(f"\n {len(models)} models trained and saved")
print(f"All models saved to: results/")
print(f"Performance report: results/classical_performance.txt")

gc.collect()
```
